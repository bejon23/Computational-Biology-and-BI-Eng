# কনসেপ্ট গাইড: কোনটা কী মাপে, কোন সিদ্ধান্তে কাজে লাগে

এই গাইডে RNA-seq analysis আর AI-output যাচাইয়ের মূল ধারণাগুলো সহজ গল্প দিয়ে বোঝানো হয়েছে। প্রতিটা ধারণায় তিনটা জিনিস আছে: গল্প, কী মাপে বা ধরে, আর কোন সিদ্ধান্তে কাজে লাগে।

---

## ১. RNA-seq
**গল্প:** একটা বিশাল কারখানায় ২০,০০০ টা বাল্ব আছে (এক একটা বাল্ব = এক একটা জিন)। RNA-seq হলো সব বাল্বের উজ্জ্বলতা একসাথে মাপা।

**কী মাপে:** কোন জিন কতটা সক্রিয়।

**ব্যবহার:** রোগীর কোষে কোন জিন বেশি বা কম কাজ করছে তা দেখে রোগের কারণ বা ওষুধের লক্ষ্য খোঁজা।

---

## ২. DE gene (Differentially Expressed gene)
**গল্প:** রোগীদের কারখানায় ৫ নম্বর বাল্ব খুব উজ্জ্বল, সুস্থদের কারখানায় নিভু নিভু। এই বাল্বটাই "DE gene"।

**সিদ্ধান্ত:** এই জিনটা রোগের সাথে জড়িত কি না, biomarker বা drug target হতে পারে কি না।

---

## ৩. Batch effect
**গল্প:** তুমি সোমবার একটা দোকানে আর মঙ্গলবার আরেকটা দোকানে আম ওজন করলে। দুই দোকানের দাঁড়িপাল্লা আলাদা। ফলে "মঙ্গলবারের আম ভারী" মনে হলো, অথচ আসলে দাঁড়িপাল্লার দোষ।

**কী ধরে:** PCA plot-এ নমুনাগুলো batch অনুযায়ী রং করো। রোগী/সুস্থ না মিশে batch অনুযায়ী আলাদা গুচ্ছ হলে batch effect আছে।

**সিদ্ধান্ত:** model-এ batch কে একটা variable হিসেবে ধরো, যাতে মেশিনের পার্থক্য আর জীববিজ্ঞানের পার্থক্য আলাদা হয়।

---

## ৪. Confounding
**গল্প:** সব রোগীর নমুনা সোমবার মেশিন-A তে, সব সুস্থের নমুনা মঙ্গলবার মেশিন-B তে। এখন পার্থক্য দেখা গেলে বোঝার উপায় নেই এটা রোগের কারণে না মেশিনের কারণে। দুটো পুরো জড়িয়ে গেছে।

**সিদ্ধান্ত:** কোনো statistics দিয়ে এটা সারে না। পরীক্ষাটা নতুন করে design করতে হবে, প্রতি batch-এ রোগী ও সুস্থ দুই ধরনের নমুনা রেখে।

**মনে রাখো:** Batch effect সামলানো যায়, কিন্তু full confounding সামলানো যায় না।

---

## ৫. p-value আর FDR
**গল্প:** তুমি ১০,০০০ বার মুদ্রা ছুঁড়ে দেখছ। কিছু ক্ষেত্রে ভাগ্যের জোরে টানা ১০ বার হেড আসবেই। ১০,০০০ জিন পরীক্ষা করলেও ভাগ্যে কিছু জিন "আলাদা" দেখাবে।

**কী মাপে:**
- p-value বলে: একটা জিন শুধু ভাগ্যে এমন দেখানোর সম্ভাবনা কত।
- FDR (Benjamini-Hochberg) বলে: আমার ঘোষিত আবিষ্কারের মধ্যে কত শতাংশ ভুয়া হতে পারে।

**প্রমাণ:** আসল কোনো signal নেই এমন data-তে raw p-value দিয়ে ৭৯টা ভুয়া জিন পাওয়া গিয়েছিল, FDR দিয়ে ০।

**সিদ্ধান্ত:** হাজার হাজার জিনে সবসময় FDR-adjusted p-value দিয়ে জিন বাছাই করো।

## ৬. TPM বনাম raw counts
**গল্প:** TPM হলো "এই ক্লাসের ভেতরে কে কার চেয়ে লম্বা" তা দেখার মাপ। কিন্তু দুই ক্লাসের মধ্যে তুলনা করতে হলে মূল মাপ (raw counts) লাগে, আর তার জন্য বিশেষ পরীক্ষা (DESeq2/edgeR) আছে যা count-এর স্বভাব বোঝে।

**সিদ্ধান্ত:**
- এক নমুনার ভেতরে জিন তুলনা বা heatmap/visualization: TPM চলে।
- রোগী বনাম সুস্থ DE পরীক্ষা: raw counts + DESeq2/edgeR।

## ৭. Ti/Tv ratio
**গল্প:** DNA-র পরিবর্তন দুই ধরনের: transition (সমগোত্রীয় বদল, যেমন A↔G) আর transversion (ভিন্ন গোত্রের বদল)। প্রকৃতিতে transition বেশি ঘটে। কিন্তু যন্ত্রের ভুল এলোমেলো, তাই দুই ধরনই প্রায় সমান দেখায়।

**কী মাপে:** Variant list-এর মান। আসল ratio whole genome-এ প্রায় ২.১, exome-এ প্রায় ৩।

**সিদ্ধান্ত:** ratio অনেক কম হলে (১ এর কাছাকাছি) ধরে নাও অনেক ভুয়া variant ঢুকেছে। তখন filter কড়া করো বা pipeline ঠিক করো।

---

## ৮. Data leakage
**গল্প:** পরীক্ষার আগে শিক্ষক প্রশ্নপত্র ফাঁস করে দিলেন। তুমি ৯০% পেলে, কিন্তু আসলে কিছু শেখোনি। Machine learning-এ test data-র তথ্য training-এর সময় model-এর কাছে চলে গেলে এমন হয়।

**ধরার উপায়:** এলোমেলো (random) data দিলেও accuracy ৯০% এর বেশি দেখালে বুঝবে leak আছে।

**সিদ্ধান্ত:** Normalization, feature selection সহ সবকিছু শুধু training data থেকে শেখো, তারপর test-এ প্রয়োগ করো। একই রোগীর নমুনা train ও test দুই দিকে রেখো না।

---

## ৯. Pseudoreplication
**গল্প:** তুমি জানতে চাও বাংলাদেশের মানুষ কী পছন্দ করে। কিন্তু একজন মানুষকেই ১০০০ বার জিজ্ঞেস করে "১০০০ জনের মত" বললে সেটা প্রতারণা। Single-cell-এ এক রোগীর ১০০০ কোষ মানে ১ জন রোগী, ১০০০ জন নয়।

**ফল:** কোষগুলোকে স্বাধীন ধরলে p-value অযথা ছোট আসে, ফলে ভুয়া আবিষ্কার।

**সিদ্ধান্ত:** প্রতি রোগীর কোষ যোগ করে pseudobulk বানাও, তারপর রোগী বনাম সুস্থ তুলনা করো।

## ১০. Hallucination
**গল্প:** AI কে জিজ্ঞেস করলে সে আত্মবিশ্বাসের সাথে বলল "RNAseqMagic নামের টুল ব্যবহার করো"। এমন কোনো টুল নেই।

**যাচাই:** টুলের নাম, flag, version সত্যিই আছে কি না ডকুমেন্টেশনে দেখো, আর code টা চালিয়ে দেখো।

**সিদ্ধান্ত:** AI-র কথা বিনা যাচাইয়ে মেনো না। উত্তর ঠিক কি না জানতে এমন data-তে চালাও যার সঠিক ফল তুমি আগে থেকে জানো।

## ১১. Benchmark
**গল্প:** AI কে মাপার পরীক্ষা, যেমন স্কুলের প্রশ্নপত্র।

**ভালো প্রশ্নের গুণ:**
- একটাই সঠিক উত্তর
- উত্তর পেতে ভাবতে হয়
- বাস্তবসম্মত data
- স্বয়ংক্রিয় নম্বর
- reference solution
- সম্ভাব্য ভুল উত্তরের তালিকা

**ব্যবহার:** একাধিক AI model এর মধ্যে কোনটা জীববিজ্ঞানে বেশি নির্ভরযোগ্য তা তুলনা করা।

## ১২. Reproducibility
**গল্প:** তুমি রান্নার রেসিপি লিখলে, কিন্তু কত গ্রাম আর কত আঁচ লিখলে না। অন্যজন একই স্বাদ পাবে না। Analysis-এও একই।

**কী করো:** Fixed seed, tool version সংরক্ষণ, Snakemake workflow, file fingerprint (checksum)। তারপর দুইবার চালিয়ে ফল মেলাও।

**সিদ্ধান্ত:** ফল মিললে অন্যরা তোমার কাজে ভরসা করতে পারে।
